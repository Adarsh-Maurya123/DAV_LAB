# Program 3: Data Visualization using Matplotlib and Seaborn

## Aim
To visualize student performance data using:
- Histograms
- Box plots
- Scatter plots
- Pair plots
- Correlation heatmaps

The dataset used is `student_performance_dataset.csv`.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("Libraries imported successfully.")
```

## 1. Load the Dataset

```python
df = pd.read_csv("student_performance_dataset.csv")
print(df.head())
```

## 2. Histogram

A histogram shows the distribution of a numerical variable.

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Marks", bins=6)
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

### Observation
The histogram shows how student marks are distributed.

## 3. Box Plot

A box plot shows the median, spread, and possible outliers.

```python
df["Study_Level"] = pd.cut(
    df["Study_Hours"], bins=[0,3,5,10],
    labels=["Low","Medium","High"]
)

plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Study_Level", y="Marks")
plt.title("Marks by Study Level")
plt.show()
```

### Observation
The box plot compares the distribution of marks for different study-hour levels.

## 4. Scatter Plot

A scatter plot shows the relationship between two numerical variables.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Study_Hours", y="Marks", s=80)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

### Observation
The sample shows a generally positive relationship between study hours and marks.

## 5. Pair Plot

A pair plot displays relationships among multiple numerical variables.

```python
numerical_columns = [
    "Study_Hours", "Attendance",
    "Assignment_Score", "Marks"
]

sns.pairplot(df[numerical_columns])
plt.show()
```

### Observation
The pair plot helps compare relationships among study hours, attendance, assignment scores, and marks.

## 6. Correlation Heatmap

A heatmap represents correlation values between numerical variables. Values range from -1 to +1.

```python
correlation_matrix = df[numerical_columns].corr()
print(correlation_matrix.round(2))

plt.figure(figsize=(8,6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)
plt.title("Correlation Heatmap")
plt.show()
```

## Conclusion

The experiment demonstrates:
- **Histogram** → data distribution
- **Box plot** → spread and possible outliers
- **Scatter plot** → relationship between two variables
- **Pair plot** → multiple pairwise relationships
- **Correlation heatmap** → correlation between numerical variables

The same student performance dataset is used for all visualizations.
